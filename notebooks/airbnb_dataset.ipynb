```python
# function for checking outliers
# It will return True if outlier is present else it will return false
def check_outlier(column: str, df: pd.DataFrame) -> bool:
        
    Q1 = df[column].quantile(0.25)
    Q2 = df[column].median()
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (IQR * 1.5)
    upper_limit = Q3 + (IQR * 1.5)

    return ((df[column] < lower_limit) | (df[column] > upper_limit)).any()
```

```python
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes == "float64":
        if check_outlier(column=column, df=df):
            df[column] = df[column].replace(np.nan, df[column].median())
        else:
            df[column] = df[column].replace(np.nan, df[column].mean())
```
