# 4. Columns transformation

```python
df.columns
```

```python
# Drop columns which is not important
columns_to_drop = ['reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license', 'country code', 'last review']

df.drop(columns=columns_to_drop, inplace=True)
print("Total columns now : ", df.shape[1])
```

```python
print("Total columns : ", df.shape[1])
```

```python
# Replacing special characters
special_chars = "!@#$%^&*()+-=[]{}|;':\",./<>?`~\\"

for chars in special_chars:
    df.columns = df.columns.str.replace(
        chars, ""
    )

# Updating columns letter cases
df.columns = df.columns.str.lower()
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ", "_")
df.columns
```

# 5 Handling Duplicates

```python
# Exploring duplicates
"""
- We have to drop duplicates based on our goal.
- If duplicates rows gives meaningful insights then we dont have to drop those rows.
- Otherwise it will give us wrong insights.

"""

print("Total rows : ", df.shape[0])
print("Total duplicate rows : ", df.duplicated().sum())
print("Total unique rows : ", len(df)-df.duplicated().sum())
print("Duplicate percentage : ", np.trunc((df.duplicated().sum()/len(df))*100), "%")
```

# 6. Handle Missing Values (NaN)

```python
print("Total null values : \n", df.isnull().sum(), "\n")
print("Null percentage/column : ")
(df.isnull().sum()/len(df))*100
```

```python
# Getting the int and float type columns

for column in df.columns:
    if df[column].dtype == "int64" or df[column].dtype == "float64":
        print(column, "=",df[column].dtype, "=",df[column].isnull().sum())
    
```

```python
# Column => df["lat"]

print("Mean : ",np.trunc(df["lat"].mean()))
print("Median : ", np.trunc(df["lat"].median()))

difference = df["lat"].mean() - df["lat"].median()
print("Differnce : ", np.trunc(difference))

df["lat"] = df["lat"].replace(np.nan, df["lat"].mean())
df["lat"].isnull().sum()
```

```python
# Column => df["long"]

print("Mean : ",np.trunc(df["long"].mean()))
print("Median : ", np.trunc(df["long"].median()))

difference = df["long"].mean() - df["long"].median()
print("Differnce : ", np.trunc(difference))


df["long"] = df["long"].replace(np.nan, df["long"].mean())
df["long"].isnull().sum()
```

```python
# Column => df["construction_year"]
print("Mean : ",np.trunc(df["construction_year"].mean()))
print("Median : ", np.trunc(df["construction_year"].median()))

difference = df["construction_year"].mean() - df["construction_year"].median()
print("Differnce : ", np.trunc(difference))

df["construction_year"] = df["construction_year"].replace(np.nan, df["construction_year"].mean())
df["construction_year"].isnull().sum()
```

```python
# Column => df["minimum_nights"]
print("Mean : ",np.trunc(df["minimum_nights"].mean()))
print("Median : ", np.trunc(df["minimum_nights"].median()))

difference = df["minimum_nights"].mean() - df["minimum_nights"].median()
print("Differnce : ", np.trunc(difference))

df["minimum_nights"] = df["minimum_nights"].replace(np.nan, df["minimum_nights"].median())
df["minimum_nights"].isnull().sum()
```

```python
# Column = df["number_of_reviews"]

print("Mean : ",np.trunc(df["number_of_reviews"].mean()))
print("Median : ", np.trunc(df["number_of_reviews"].median()))

difference = df["number_of_reviews"].mean() - df["number_of_reviews"].median()
print("Differnce : ", np.trunc(difference))

print("Maximum value : ", df["number_of_reviews"].max())
print("Minimum value : ", df["number_of_reviews"].min())

df["number_of_reviews"] = df["number_of_reviews"].replace(np.nan, df["number_of_reviews"].median())
df["number_of_reviews"].isnull().sum()
```

```python
# printing catagorical column
for column in df.columns:
    if df[column].dtype == "object":
        if df[column].isnull().sum()>0:
            missing_pct = np.trunc((df[column].isnull().sum()/len(df[column]))*100)
            if missing_pct <= 10:
                df[column] = df[column].replace(np.nan, df[column].mode()[0])
```

```python
print("Total null values now : \n", df.isnull().sum())
```
